```python
# Infer schema field name from the header row
df = (spark.read
       .option("headerRows", 1)
       .excel("/Volumes/datasciencebasics/youtube/gpt4all/test.xlsx"))
display(df)
```

```python
# Infer schema field name from the header row
df = (spark.read
       .option("headerRows", 1)
       .excel("/Volumes/datasciencebasics/youtube/gpt4all/test_complex.xlsx"))
display(df)
```

```python
df = (spark.read.format("excel")
       .option("headerRows", 1)
       .option("dataAddress", "Sheet3!C7:D9")
       .load("/Volumes/datasciencebasics/youtube/gpt4all/test_complex.xlsx"))
display(df)
```

## Reading as table

```python
%sql
-- Read an entire Excel file with explicit schema
CREATE OR REPLACE TABLE my_table AS
SELECT * FROM read_files(
  "/Volumes/datasciencebasics/youtube/gpt4all/test.xlsx",
  headerRows => 1,
  schemaEvolutionMode => "none"
);

select * from my_table
```

```python
# list sheets
df = (spark.read.table("datasciencebasics.youtube.test_yt"))
display(df)
```

## Writing to Excel

```python
# Write DataFrame to Excel file with header
# This will save the DataFrame 'df' to an Excel file in the specified location with column headers

df.write.format("excel") \
  .option("headerRows", 1) \
  .save("/Volumes/datasciencebasics/youtube/gpt4all/output_yt.xlsx")
```
